# CardioScore PyMEA — self-downloading held-out validation

This notebook requires **no manual dataset upload** for PREPARE. It downloads the complete Cardio PyMEA release (DOI `10.5068/D14H5C`) from Zenodo record `6524838`, verifies the published MD5 checksums for all four files, caches them in Google Drive, runs validation preflight tests, generates blinded windows, and automatically downloads the annotation ZIP.

After annotation, switch `MODE` to `SCORE`. Upload only the annotation-package ZIP plus one or two completed annotation CSVs. The workflow automatically downloads the final results ZIP.

If the exact historical 20-window development `sheet_manifest.json` is present anywhere in your Google Drive, the workflow finds it automatically. Otherwise it reconstructs the historical default exclusion batch and marks the result provisional/exploratory rather than confirmatory.


```python
#@title 1) Choose workflow mode
import os
MODE = "PREPARE"  #@param ["PREPARE", "SCORE"]
os.environ["CARDIOSCORE_PYMEA_MODE"] = MODE
os.environ["CARDIOSCORE_PYMEA_N_WINDOWS"] = "40"
os.environ["CARDIOSCORE_PYMEA_WINDOW_S"] = "10"
os.environ["CARDIOSCORE_PYMEA_HELDOUT_SEED"] = "20261019"
print("Mode:", MODE)
```

```python
#@title 2) Run SHA-pinned Cardio PyMEA workflow
import hashlib, urllib.request

WORKFLOW_SHA = "d698409878a9eba30c742e19dbc47c57fa11091a"
WORKFLOW_URL = (
    "https://raw.githubusercontent.com/Virelion-Biotech/Virelion-CardioScore/"
    f"{WORKFLOW_SHA}/scripts/validation/run_pymea_colab_workflow.py"
)
source = urllib.request.urlopen(WORKFLOW_URL).read()
print("Workflow SHA:", WORKFLOW_SHA)
print("Workflow source SHA-256:", hashlib.sha256(source).hexdigest())
exec(compile(source, "run_pymea_colab_workflow.py", "exec"), {"__name__": "__main__"})
```

## What happens next

### PREPARE
The notebook downloads and checksum-verifies:
- `Day9__Nov1_130pm.txt`
- `File_Parameter_List.xlsx`
- `README_cardio_pymea.txt`
- `untreated_MEA120_only.xlsx`

Then it automatically downloads `cardioscore_pymea_annotation_package_20261019.zip`. Annotate the PNGs blindly and save the completed CSV as `annotator_A.csv`; a second independent annotator is preferred.

### SCORE
Change `MODE` to `SCORE`, rerun both code cells, and upload the annotation package plus the completed CSV file(s). The notebook automatically downloads `cardioscore_pymea_heldout_validation_results.zip`. Send that ZIP back to ChatGPT.
